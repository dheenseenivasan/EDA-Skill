# Superstore Dataset Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
```

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
```

```python

df.head()
```

```python

category_sales = df.groupby('Category')['Sales'].sum()

category_sales
```

```python

plt.figure(figsize=(8, 5))

category_sales.plot(kind='bar')

plt.title('Total Sales for Each Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(y=df['Profit'])

plt.title('Profit Distribution')
plt.ylabel('Profit')

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.scatter(df['Discount'], df['Profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')

plt.show()
```

```python

numerical_columns = df.select_dtypes(include='number')

correlation_matrix = numerical_columns.corr()

correlation_matrix
```

```python

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix Heatmap')

plt.show()
```

```python

category_profit = df.groupby('Category')['Profit'].sum()

category_profit
```

```python

highest_profit_category = category_profit.idxmax()
highest_profit = category_profit.max()

print("Category with Highest Total Profit:", highest_profit_category)
print("Highest Total Profit:", highest_profit)
```

```python

highest_sales_category = category_sales.idxmax()
correlation = df['Discount'].corr(df['Profit'])

print("Business Insights:")
print()

print("1. The category with the highest total profit is",
      highest_profit_category + ".")

print("2. The category with the highest total sales is",
      highest_sales_category + ".")

print("3. The correlation between Discount and Profit is",
      round(correlation, 2), ".")
```
